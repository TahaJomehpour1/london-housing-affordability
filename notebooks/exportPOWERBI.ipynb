# 05 Export for Power BI

Power BI runs in a Windows VM, so it reads CSV exports rather than connecting to MySQL directly.
The views hold all the business logic. This notebook only copies their output to files.

Outputs go to `data/powerbi/`, which is not committed to GitHub because of file size.

```python
from getpass import getpass
from pathlib import Path

import pandas as pd
from sqlalchemy import create_engine

password = getpass("MySQL password: ")
engine = create_engine(f"mysql+pymysql://root:{password}@localhost:3306/london_property")

out_dir = Path("../data/powerbi")
out_dir.mkdir(parents=True, exist_ok=True)
```

## Export each table

`sales_v` becomes the fact table. The other three are already summarised, so they are small.

```python
exports = {
    "fact_sales": "SELECT * FROM sales_v",
    "affordability": "SELECT * FROM affordability_v",
    "repayments": "SELECT * FROM repayment_v",
    "rates": "SELECT * FROM rates",
}

for name, query in exports.items():
    df = pd.read_sql(query, engine)
    df.to_csv(out_dir / f"{name}.csv", index=False)
    print(f"{name}: {len(df):,} rows, {df.shape[1]} columns")
```
